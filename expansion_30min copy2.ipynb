```python
import leafmap.foliumap as leafmap

POINT_N_VIEW = 8145
DAY_BEFORE_VIEW = "20260611"
DAY_AFTER_VIEW = "20260612"

gdf_day_point = gdf_tt_extended[gdf_tt_extended["N"] == POINT_N_VIEW]
gdf_day_zone_before = filter_zones_by_moment(
    df=gdf_tt_zones_30m_extended[gdf_tt_zones_30m_extended["N"] == POINT_N_VIEW],
    day=DAY_BEFORE_VIEW,
    dissolve=True,
)
gdf_day_zone_after = filter_zones_by_moment(
    df=gdf_tt_zones_30m_extended[gdf_tt_zones_30m_extended["N"] == POINT_N_VIEW],
    day=DAY_AFTER_VIEW,
    dissolve=True,
)

m = leafmap.Map()
m.add_gdf(
    gdf_day_zone_after,
    layer_name="Зона после",
    style={"color": "blue", "fillColor": "blue", "fillOpacity": 0.3},
)
m.add_gdf(
    gdf_day_zone_before,
    layer_name="Зона до",
    style={"color": "red", "fillColor": "red", "fillOpacity": 0.3},
)
m.add_gdf(
    gdf_day_point[["N", "id_tt", "name_TT", "Тип", "geometry"]],
    layer_name="ТТ",
    point_style={
        "radius": 5,
        "color": "green",
        "fillColor": "green",
        "fillOpacity": 1,
    },
)
m.zoom_to_gdf(gdf_day_zone_after)
m.add_layer_control()
m
```
